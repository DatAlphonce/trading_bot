```python
# Load API keys from the .env file so they never appear in the code for security purposes
import os
from dotenv import load_dotenv

load_dotenv()  # reads .env into environment variables

api_key = os.getenv("API_KEY")
api_secret = os.getenv("API_SECRET")
```

```python
# Import the python-binance and time packages
from binance.client import Client
import time
```

```python
# connect to the Binance API using the API keys
client = Client(api_key,api_secret,testnet=True)
```

```python
# get account information to check the testnet connection and API keys are working correctly
client.get_account()
```

```python
# Set the trading strategy parameters
symbol = 'BTCUSDT'
buy_price_threshold = 80000
sell_price_threshold = 83256
trade_quantity = 0.001
```

```python
# Price helper function to get the current price of the symbol chosen for trading
def get_current_price(symbol):
    ticker = client.get_symbol_ticker(symbol=symbol)
    return float(ticker['price'])
```

```python
# get the current price of the symbol chosen for trading
get_current_price(symbol)
```

```python
# Buy order function
def place_buy_order(symbol,quantity):
    order = client.order_market_buy(symbol=symbol, quantity=quantity)
    print(f"Buy order done: {order}")
```

```python
# Manual test for the buy order function
place_buy_order(symbol,trade_quantity)
```

```python
# Sell order function
def place_sell_order(symbol,quantity):
    order = client.order_market_sell(symbol=symbol, quantity=quantity)
    print(f"Sell order done: {order}")
```

```python
# Smoke test of the sell order function
place_sell_order(symbol,trade_quantity)
```

```python
# Main bot loop
def trading_bot():
    in_position = False

    while True:
        current_price = get_current_price(symbol)
        print(f"Current price of {symbol}: {current_price}")

        if not in_position:
            if current_price < buy_price_threshold: # Not in a position: buy if price drops below threshold
                print(f"Price is below {buy_price_threshold}. Placing buy order.")
                place_buy_order(symbol, trade_quantity)
                in_position = True
        else:
            if current_price > sell_price_threshold: # In a position: sell if price rises above threshold
                print(f"Price is above {sell_price_threshold}. Placing sell order.")
                place_sell_order(symbol, trade_quantity)
                in_position = False

        time.sleep(3)
```

```python
if __name__ == "__main__":
    trading_bot()
```
